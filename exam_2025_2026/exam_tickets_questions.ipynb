```python
import random
```

```python
def generate_tickets_markdown(questions, num_tickets, questions_per_ticket):
    """
    Generates random tickets from a list of questions and outputs them in formatted Markdown.
    
    Args:
        questions (list): A list of strings representing the available questions.
        num_tickets (int): The total number of tickets to generate.
        questions_per_ticket (int): The number of unique questions each ticket should contain.
        
    Returns:
        str: A beautifully formatted Markdown string containing all the tickets.
    """
    # Safety check: We can't pick more unique questions per ticket than we actually have
    if questions_per_ticket > len(questions):
        raise ValueError("Questions per ticket cannot exceed the total number of available questions.")

    # Start building our Markdown string
    md_output = "# 📝 Examination Tickets\n\n"

    for ticket_num in range(1, num_tickets + 1):
        # random.sample picks unique elements from the list without replacement
        selected_questions = random.sample(questions, questions_per_ticket)
        
        # Add the Ticket header
        md_output += f"## Ticket {ticket_num}\n\n"
        
        # Add the numbered questions
        for i, question in enumerate(selected_questions, 1):
            md_output += f"{i}. {question}\n"
            
        # Add a visual divider between tickets
        md_output += "\n---\n\n"

    return md_output    
```
